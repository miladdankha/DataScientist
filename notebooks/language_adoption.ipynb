# Language adoption and curriculum planning

Among professionals living in France who want to use a language next year, how many reported using it last year? This distinction helps a French engineering school assess the case for an introductory lab or a follow-on project. Source: [Stack Overflow Developer Survey 2026](https://survey.stackoverflow.co/2026/).

```python
from pathlib import Path
import sys

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
sys.path.insert(0, str(root))

from get_data import ANSWERS, fetch
from logic import load_survey, population, language_adoption_counts

if not ANSWERS.is_file():
    fetch()
group = population(load_survey(ANSWERS), "France")
counts, paired = language_adoption_counts(group)
print(f"{len(group)} professional respondents in France; {paired} answered both language items.")
```

## Trace the denominator

The overview includes everyone who answered the desired-use question. Classifying someone by past use also requires a usable answer to the past-use question. A missing answer gives us no evidence about prior use.

```python
language = "Rust"
desired = group["LanguagesWantToWorkWith"].dropna().str.split(";")
overview_count = sum(language in answers for answers in desired)
split = counts[language]
print(f"Want {language}: {overview_count}; with both answers: {sum(split.values())}.")
for category, n in split.items():
    print(f"{category}: {n} ({n / sum(split.values()):.1%})")
```

## Compare two teaching decisions

```python
for language in ("Rust", "Python"):
    split = counts[language]
    total = sum(split.values())
    prospective = split["No reported past-year use"]
    print(f"{language}: {prospective}/{total} ({prospective/total:.1%}) wanted the language without reporting past-year use.")
```

In this sample, almost half of interested Rust respondents reported no past-year use, compared with about one in ten for Python. This supports exploring an introductory Rust lab and a follow-on Python project. Before deciding, ask incoming students about their experience. These are voluntary responses from professionals, and someone may have used a language before the survey’s one-year window.
